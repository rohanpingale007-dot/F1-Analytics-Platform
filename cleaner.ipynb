USING PANDAS TO IMPORT DATA INTO PYTHON

```python
import pandas as pd 
import numpy as np
import os

# COMMON DATA LOADING

circuits=pd.read_csv(r'D:\Game\F1\data\circuits.csv')
drivers=pd.read_csv(r'D:\Game\F1\data\drivers.csv')
races = pd.read_csv(r'D:\Game\F1\data\races.csv')
seasons=pd.read_csv(r'D:\Game\F1\data\seasons.csv')

#CONST DATA LOADING
constructors=pd.read_csv(r'D:\Game\F1\data\const\constructors.csv')
constructor_results=pd.read_csv(r'D:\Game\F1\data\const\constructor_results.csv')
constructor_standings=pd.read_csv(r'D:\Game\F1\data\const\constructor_standings.csv')

# result data loading

results=pd.read_csv(r'D:\Game\F1\data\result\results.csv')
driver_standings=pd.read_csv(r'D:\Game\F1\data\result\driver_standings.csv')
laptimes=pd.read_csv(r'D:\Game\F1\data\result\lap_times.csv')
pitstops=pd.read_csv(r'D:\Game\F1\data\result\pit_stops.csv')
qualifying=pd.read_csv(r'D:\Game\F1\data\result\qualifying.csv')
sprintings=pd.read_csv(r'D:\Game\F1\data\result\sprint_results.csv')
status=pd.read_csv(r'D:\Game\F1\data\result\status.csv')
```

```python
os.makedirs(r'D:\Game\F1\data\cleaned',exist_ok=True)
```

```python
def clean_101(df):

    #remove dups
    df=df.drop_duplicates()

    #removing extra spaces
    df.columns = df.columns.str.strip()

    #replace empty strings with NaN
    df=df.replace('', np.nan, inplace=True)
    return df
```

using the clean101 function to clean every dataset with clean101


```python
drivers=clean_101(drivers)

drivers['dob']=pd.to_datetime(drivers['dob'],errors='coerce')

drivers.to_csv(r'D:\Game\F1\data\cleaned\drivers.csv',index=False)
```

```python
races=clean_101(races)

races['date']=pd.to_datetime(races['date'],errors='coerce')

races.to_csv(r'D:\Game\F1\data\cleaned\races.csv',index=False)
```

```python
results=clean_101(results)

results['milliseconds']=pd.to_numeric(results['milliseconds'],errors='coerce')

results['points']=pd.to_numeric(results['points'],errors='coerce').fillna(0)

results.columns=results.columns.str.replace(r'\N','DNF')
results.to_csv(r'D:\Game\F1\data\cleaned\results.csv',index=False)

```

```python
pitstops=clean_101(pitstops)

pitstops['milliseconds']=pd.to_numeric(pitstops['milliseconds'],errors='coerce')

pitstops.to_csv(r'D:\Game\F1\data\cleaned\pitstops.csv',index=False)
```

```python
remaining_files={
    'circuits':circuits,
    'seasons':seasons,
    'constructors':constructors,
    'constructor_results':constructor_results,
    'constructor_standings':constructor_standings,
    'driver_standings':driver_standings,
    'qualifying':qualifying,
    'seasons':seasons,
    'sprintings':sprintings,
    'status':status
    }

for name,df in remaining_files.items():
    df=clean_101(df)
    df.to_csv(rf'D:\Game\F1\data\cleaned\{name}.csv',index=False)
```

```python
laptimes=clean_101(laptimes)

laptimes['milliseconds']=pd.to_numeric(laptimes['milliseconds'],errors='coerce')
laptimes.to_csv(r'D:\Game\F1\data\cleaned\laptimes.csv',index=False)
```
