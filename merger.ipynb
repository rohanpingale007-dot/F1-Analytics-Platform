```python
import pandas as pd 
import numpy as np
circuits=pd.read_csv(r'D:\Game\F1\data\cleaned\circuits.csv')
drivers=pd.read_csv(r'D:\Game\F1\data\cleaned\drivers.csv')
race= pd.read_csv(r'D:\Game\F1\data\cleaned\races.csv')
seasons=pd.read_csv(r'D:\Game\F1\data\cleaned\seasons.csv')
constructors=pd.read_csv(r'D:\Game\F1\data\cleaned\constructors.csv')
constructor_results=pd.read_csv(r'D:\Game\F1\data\cleaned\constructor_results.csv')
constructor_standings=pd.read_csv(r'D:\Game\F1\data\cleaned\constructor_standings.csv')
results=pd.read_csv(r'D:\Game\F1\data\cleaned\results.csv')
driver_standings=pd.read_csv(r'D:\Game\F1\data\cleaned\driver_standings.csv')
laptimes=pd.read_csv(r'D:\Game\F1\data\cleaned\laptimes.csv')
pitstops=pd.read_csv(r'D:\Game\F1\data\cleaned\pitstops.csv')
qualifying=pd.read_csv(r'D:\Game\F1\data\cleaned\qualifying.csv')
sprintings=pd.read_csv(r'D:\Game\F1\data\cleaned\sprintings.csv')
status=pd.read_csv(r'D:\Game\F1\data\cleaned\status.csv')
```

```python
import pandas as pd 
import numpy as np
race=race.rename( columns={
    'name':'Race Name',
    'year':'Season',
    'date':'Race Date'})
circuits=circuits.rename(columns={
    'name':'Circuit Name',
    'country': 'Country'})
constructors =constructors.rename(columns={
    'name':'Consuctor Team'})
drivers['Driver Name']=drivers['forename']+' '+drivers['surname']
```

```python
master=results.merge(
    race[['raceId','Race Name','Season','circuitId','Race Date']],
    on='raceId',
    how='left')
```

```python
master=master.merge(
    circuits[['circuitId','Circuit Name','Country']],
    on='circuitId',
    how='left')
```

```python
master=master.merge(
    drivers[['driverId','Driver Name','nationality']],
    on='driverId',
    how='left')
```

```python
master=master.merge(
    constructors[['constructorId','Consuctor Team']],
    on='constructorId',
    how='left')
```

```python
pitstops=pitstops.rename(columns={'stop':'Number Of Pit-Stops'})
pitstops
```

```python
pit_summary = (
    pitstops
    .groupby(["raceId", "driverId"])
    .size()
    .reset_index(name="Number Of Pit-Stops"))
```

```python
master = master.merge(
    pit_summary,
    on=["raceId", "driverId"],
    how="left")
```

```python
master["Number Of Pit-Stops"] = (
    master["Number Of Pit-Stops"]
    .fillna(0)
    .astype(int))
```

```python
status = status.rename(columns={
    "status": "Status"})
```

```python
master = master.merge(
    status[["statusId", "Status"]],
    on="statusId",
    how="left")
```

```python
master.rename(columns={
    "grid": "Grid Position",
    "positionOrder": "Finish Position",
    "points": "Points",
    "laps": "Laps",
    "time": "Race Time",
    "fastestLap": "Fastest Lap",
    "nationality": "Nationality",
    "Consuctor Team": "Constructor"
}, inplace=True)
```

```python
front_cols = [
    "Season",
    "Race Name",
    "Race Date",
    "Circuit Name",
    "Country",
    "Driver Name",
    "Nationality",
    "Constructor",
    "Grid Position",
    "Finish Position",
    "Points",
    "Fastest Lap",
    "Number Of Pit-Stops",
    "Status",
    "Laps",
    "Race Time"
]

master = master[front_cols + [col for col in master.columns if col not in front_cols]]
```

```python
master.to_csv(r'D:\Game\F1\merged\'masterF1.csv')
```
